```python
df=pd.read_csv("/content/drive/MyDrive/level.csv")
df.head()
```

```python
df.drop(columns=["Unnamed: 2","Salary"],inplace=True)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
```

```python
x=df.iloc[:,:-1]
y=df["Salary1"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
dt=DecisionTreeRegressor(criterion="squared_error",max_depth=9,splitter="random")
dt.fit(x_train,y_train)
```

```python
dt.score(x_train,y_train)*100,dt.score(x_test,y_test)*100
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
dt={"criterion":["squared_error","friedman_mse","absolute_error","poisson"],
    "splitter":["best","random"],
    "max_depth":[i for i in range(2,20)]}
```

```python
gd=GridSearchCV(DecisionTreeRegressor(),param_grid=dt)
gd
```

```python
gd.fit(x_train,y_train)
```

```python
gd.best_params_
```

```python
gd.best_score_
```

```python
rd=RandomizedSearchCV(DecisionTreeRegressor(),param_distributions=dt)
rd.fit(x_train,y_train)
```

```python
rd.best_params_
```

```python
rd.best_score_
```

```python
dt=DecisionTreeRegressor(criterion="squared_error",max_depth=12,splitter="best")
dt.fit(x_train,y_train)
```

```python
dt.score(x_train,y_train)*100,dt.score(x_test,y_test)*100
```
