```python
df=pd.read_csv("/content/drive/MyDrive/placement (1).csv")
df.head()
```

```python
x=df.iloc[:,:-1]
y=df["package"]
```

```python
new_df=df.head(10)
new_df
```

```python
new_x=new_df.iloc[:,:-1]
new_y=new_df["package"]
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import LeaveOneOut,LeavePOut,KFold,StratifiedKFold
```

```python
lo=LeaveOneOut()

for train,test in lo.split(new_x,new_y):
  print(train,test)
```

```python
lo=LeavePOut(p=2)

for train,test in lo.split(new_x,new_y):
  print(train,test)
```

```python
lo=KFold(n_splits=5)

for train,test in lo.split(new_x,new_y):
  print(train,test)
```

```python
lo=StratifiedKFold() # Only work on Classification Analysis

for train,test in lo.split(new_x,new_y):
  print(train,test)
```

```python
from sklearn.linear_model import LinearRegression
```
