```python
import pandas as pd

df=pd.read_csv('/content/healthcare_anomaly_detection_100_rows.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
```

```python
clf=IsolationForest(contamination="auto")
clf.fit(df)
prediction=clf.predict(df)
```

```python
prediction
```

```python
import numpy as np
index=np.where(prediction<0)
index
```

```python
x=df.values
```

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[index,0],x[index,1],color='red')
plt.show()
```
